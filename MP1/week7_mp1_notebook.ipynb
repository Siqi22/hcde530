Answer your three research questions using pandas. Each question should have:

**Question 1:** *How is social media use associated with adolescent mental health and feelings of hopelessness?*

```python
# Question 1 — Social media indicator vs hopelessness (both coded 0/1 in this extract).
# I drop rows missing either answer so the rate compares the same kind of respondent.

q1 = df[["SocialMedia", "Hopelessness"]].dropna()
print(f"Students who answered BOTH items: {len(q1):,} (out of {len(df):,} total rows)\n")

tbl = q1.groupby("SocialMedia")["Hopelessness"].agg(["count", "mean"]).round(3)
tbl["pct_hopeless_1"] = (tbl["mean"] * 100).round(1)
print("Mean Hopelessness (= share answering 'yes' if 1 means yes) by SocialMedia code:")
print(tbl)

print("\nPlain-language rates (Hopelessness == 1):")
for code, grp in q1.groupby("SocialMedia"):
    pct = (grp["Hopelessness"] == 1).mean() * 100
    print(f"  SocialMedia == {int(code)}: {pct:.1f}% report hopelessness (n={len(grp):,})")
```

**Interpretation:** Among students who answered **both** `SocialMedia` and `Hopelessness`, the share reporting hopelessness is **higher** when the social-media risk indicator is **1** than when it is **0** (roughly **44% vs 33%** in this file). That is a **descriptive association**: it does not prove social media *caused* hopelessness—both could be driven by stress, sleep, or other factors not in this two-variable slice. 

**Question 2:** *What is the relationship between cyberbullying and adolescent mental health outcomes?*

*Same approach for Question 2: only students with non-missing answers on **both** cyberbullying and the mental-health outcome are compared.*

```python
# Question 2 — Cyberbullying vs self-reported "not good" mental health (0/1 items).

q2 = df[["CyberBullying", "NotGoodMentalHealth"]].dropna()
print(f"Students who answered BOTH items: {len(q2):,}\n")

tbl2 = q2.groupby("CyberBullying")["NotGoodMentalHealth"].agg(["count", "mean"]).round(3)
tbl2["pct_not_good_1"] = (tbl2["mean"] * 100).round(1)
print("Mean NotGoodMentalHealth by CyberBullying (1 = experienced cyberbullying in this coding):")
print(tbl2)

print("\nRates (NotGoodMentalHealth == 1):")
for code, grp in q2.groupby("CyberBullying"):
    pct = (grp["NotGoodMentalHealth"] == 1).mean() * 100
    label = "experienced cyberbullying" if int(code) == 1 else "did not"
    print(f"  {label}: {pct:.1f}% (n={len(grp):,})")
```

**Interpretation:** Students who report **cyberbullying** (`CyberBullying == 1`) also report **poor mental health** (`NotGoodMentalHealth == 1`) at a **much higher rate** than students who do not (about **51% vs 26%** in this extract). That pattern matches what we would expect from past research, but this table is still **only a cross-section**: bullying and mental health can reinforce each other over time, and the survey cannot separate direction. 

**Question 3:** *How does sleep relate to social media use and mental health among teenagers?*

```python
# Question 3 — Eight+ hours of sleep vs social media indicator and hopelessness.
# EightOrMoreHoursSleep is 1 if student reports 8+ hours sleep, 0 otherwise (per YRBS-style coding here).

q3 = df[["EightOrMoreHoursSleep", "SocialMedia", "Hopelessness"]].dropna()
print(f"Students with all three items non-missing: {len(q3):,}\n")

print("Mean SocialMedia risk indicator by sleep group (0 = under 8h, 1 = 8h or more):")
print(q3.groupby("EightOrMoreHoursSleep")["SocialMedia"].mean().round(3))
print("\nMean Hopelessness by sleep group:")
print(q3.groupby("EightOrMoreHoursSleep")["Hopelessness"].mean().round(3))

print("\nHopelessness rate (%) by sleep group:")
for code, grp in q3.groupby("EightOrMoreHoursSleep"):
    label = "8+ hours sleep" if int(code) == 1 else "Under 8 hours sleep"
    pct = (grp["Hopelessness"] == 1).mean() * 100
    print(f"  {label}: {pct:.1f}% (n={len(grp):,})")
```

**Interpretation:** Among teens who answered all three items, those reporting **8 or more hours of sleep** show **lower average hopelessness** than those under eight hours (about **27% vs 46%** reporting hopelessness). The social-media indicator is also **slightly lower** on average in the 8+ sleep group. Together that supports a **sleep–mood** story practitioners care about, but it is still observational: teens who sleep more may differ in many other ways (parent rules, homework load, anxiety).

## Section 4 — Visualization

Create **at least one chart per research question** (three charts total below). Each chart should:

- Have a **title that states the finding**, not only the variable names
- Have **labeled axes** (Plotly `labels=` helps)
- Use a chart type that fits the data (**grouped bar** for Chart 1 to compare **two outcomes** at social-media 0 vs 1; **simple bar** for Charts 2–3 for a **single proportion** across two groups)

Run the cells **in order** after `df` is loaded. Each block builds its own small summary table from the same filters used in Section 3.


### Chart 1 — Question 1 (social media, mental health, and hopelessness)

**Grouped bar chart:** at each **social-media indicator** (0 vs 1), two bars compare the **share answering “yes”** on **hopelessness** and on **not-good mental health** (both items are 0/1 here, so means equal proportions). Uses rows with **all three** columns non-missing.


```python
# --- Chart 1 — Question 1: SocialMedia vs Hopelessness and NotGoodMentalHealth ---
# Grouped bars: compare two 0/1 outcomes at social-media 0 vs 1 (pairwise-complete rows).

q1_plot = df[["SocialMedia", "Hopelessness", "NotGoodMentalHealth"]].dropna()
summary1 = q1_plot.groupby("SocialMedia", as_index=False).agg(
    Hopelessness=("Hopelessness", "mean"),
    NotGoodMentalHealth=("NotGoodMentalHealth", "mean"),
)
long1 = summary1.melt(
    id_vars="SocialMedia",
    value_vars=["Hopelessness", "NotGoodMentalHealth"],
    var_name="Outcome",
    value_name="proportion",
)
long1["Outcome"] = long1["Outcome"].map(
    {
        "Hopelessness": "Hopelessness",
        "NotGoodMentalHealth": "Not good mental health",
    }
)

fig1 = px.bar(
    long1,
    x="SocialMedia",
    y="proportion",
    color="Outcome",
    barmode="group",
    title="Hopelessness and poor mental health by social-media indicator",
    labels={
        "SocialMedia": "Social media indicator (0 vs 1)",
        "proportion": "Proportion answering yes",
        "Outcome": "Outcome",
    },
)

fig1.update_layout(
    yaxis_tickformat=".0%",
    xaxis=dict(tickmode="linear", tick0=0, dtick=1),
    legend_title_text="",
)

fig1.show()
```

### Chart 2 — Question 2 (cyberbullying and mental health)

```python
# --- Chart 2 — Question 2: cyberbullying vs self-rated poor mental health ---

q2_plot = df[["CyberBullying", "NotGoodMentalHealth"]].dropna()
summary2 = q2_plot.groupby("CyberBullying", as_index=False)["NotGoodMentalHealth"].mean()
summary2["group"] = summary2["CyberBullying"].map(
    {0.0: "No cyberbullying (0)", 1.0: "Cyberbullying (1)"}
)

fig2 = px.bar(
    summary2,
    x="group",
    y="NotGoodMentalHealth",
    title="Poor mental health is much more common among students who report cyberbullying",
    labels={
        "NotGoodMentalHealth": "Proportion with not-good mental health",
        "group": "",
    },
)
fig2.update_layout(yaxis_tickformat=".0%", showlegend=False)
fig2.show()
```

### Chart 3 — Question 3 (sleep and hopelessness)

Same filter as the printed tables in Section 3: only students with **both** `EightOrMoreHoursSleep` and `Hopelessness` answered.

```python
# --- Chart 3 — Question 3: sleep (8+ hours) vs hopelessness ---
# Pairwise complete rows: `EightOrMoreHoursSleep` and `Hopelessness` both answered (Section 3 filter).

q3_plot = df[["EightOrMoreHoursSleep", "Hopelessness"]].dropna()
summary3 = q3_plot.groupby("EightOrMoreHoursSleep", as_index=False)["Hopelessness"].mean()
summary3["group"] = summary3["EightOrMoreHoursSleep"].map(
    {0.0: "Under 8 hours sleep", 1.0: "8 or more hours sleep"}
)

fig3 = px.bar(
    summary3,
    x="group",
    y="Hopelessness",
    title="Students who sleep 8+ hours report hopelessness less often",
    labels={"Hopelessness": "Proportion reporting hopelessness", "group": ""},
)
fig3.update_layout(yaxis_tickformat=".0%", showlegend=False)
fig3.show()
```

**Chart rationale:** **Chart 1** uses **`plotly.express.bar`** with **`barmode="group"`** so readers can compare **two outcomes** (**hopelessness** and **not-good mental health**) **side by side** at **social-media 0 vs 1** without implying a continuous “trend” between the two points (which a line chart can suggest when *x* is categorical). **Charts 2 and 3** each show **one outcome proportion** across **two groups**, so a **simple bar chart** makes the **gap in height** the main story. Titles describe **patterns in the survey**, not causation—many factors co-occur with sleep, bullying, and screen-related risk, and this cross-section cannot prove what would change if one factor changed.
